```python
import random

Answer = random.randint(0, 100)

print("--- 歡迎來到猜數字遊戲！ ---")
print("我已經選好了一個 0 到 100 之間的數字。")

Guess = -1 

while Guess != Answer:
    try:
        Guess = int(input("請輸入你猜的數字："))
        
        if Guess < Answer:
            print("太小了！再試一次。")
        elif Guess > Answer:
            print("太大了！再試一次。")
        else:
            print(f"恭喜你！猜對了，答案就是 {Answer}！")
            
    except ValueError:
        print("請輸入有效的整數數字喔！")

print("遊戲結束，感謝遊玩！")
```

```python

n1 = int(input("請輸入第一個整數 n1: "))
n2 = int(input("請輸入第二個整數 n2: "))

gcd = 1
k = 2

while k <= n1 and k <= n2:
    if n1 % k == 0 and n2 % k == 0:
        gcd = k
    k += 1

print(f"數字 {n1} 與 {n2} 的最大公因數為: {gcd}")
```

```python
n = int(input("Enter the number of lines (1-15): "))

for i in range(1, n + 1):
    print(" " * (n - i) * 3, end="")

    for j in range(i, 0, -1):
        print(f"{j:3d}", end="")

    for j in range(2, i + 1):
        print(f"{j:3d}", end="")
        
    print()
```
