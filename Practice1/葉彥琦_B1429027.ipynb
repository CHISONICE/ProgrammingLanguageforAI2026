```python
CurrentPopulation = 312033422
Time = 365 * 24 * 60 * 60
BPY = Time / 7
DPY = Time / 13
IPY = Time / 45
AnnualChange = BPY-DPY+IPY

print(f"Current Population:{CurrentPopulation}")

for Year in range(1,6):
    CurrentPopulation += AnnualChange
    print(f"Year {Year} population:{int(CurrentPopulation):,}")
```

```python
import time

TotalSeconds = int(time.time())
CurrentSecond = TotalSeconds % 60
TotalMinutes = TotalSeconds // 60
CurrentMinute = TotalMinutes % 60
TotalHours = TotalMinutes // 60
CurrentHour = TotalHours % 24

print(f"Current GMT time is {CurrentHour:02d}:{CurrentMinute:02d}:{CurrentSecond:02d}")
```
