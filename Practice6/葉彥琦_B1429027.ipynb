```python
class Bear:
    def eats(self):
        return 'berries'
    
class Rabbit:
    def eats(self):
        return 'clover'

class Octothorpe:
    def eats(self):
        return 'campers'

bear = Bear()
rabbit = Rabbit()
octothorpe = Octothorpe()

print(f"Bear eats: {bear.eats()}")
print(f"Rabbit eats: {rabbit.eats()}")
print(f"Octothorpe eats: {octothorpe.eats()}")
```

```python
class Laser:
    def does(self):
        return "disintegrate"

class Claw:
    def does(self):
        return "crush"

class SmartPhone:
    def does(self):
        return "ring"

class Robot:
    def __init__(self):
        self.laser = Laser()
        self.claw = Claw()
        self.smartphone = SmartPhone()

    def does(self):
        print(f"我的雷射正在：{self.laser.does()}")
        print(f"我的爪子正在：{self.claw.does()}")
        print(f"我的手機正在：{self.smartphone.does()}")

my_robot = Robot()
my_robot.does()
```

```python
class Element:
    def __init__(self, name, symbol, number):
        self.__name = name
        self.__symbol = symbol
        self.__number = number

    @property
    def name(self):
        return self.__name

    @property
    def symbol(self):
        return self.__symbol

    @property
    def number(self):
        return self.__number

    def dump(self):
        print(f"dump() 輸出內容：{self.name}, {self.symbol}, {self.number}")

    def __str__(self):
        return f"__str__ 輸出內容：name={self.name}, symbol={self.symbol}, number={self.number}"


el_dict = {'name': 'Hydrogen', 'symbol': 'H', 'number': 1}

hydrogen = Element(**el_dict)

hydrogen.dump()

print(hydrogen)

print(f"透過 Getter 取得名稱: {hydrogen.name}")
```

```python
import math

class QuadraticEquation:
    def __init__(self, a, b, c):
        self.__a = a
        self.__b = b
        self.__c = c

    def get_a(self):
        return self.__a

    def get_b(self):
        return self.__b

    def get_c(self):
        return self.__c

    def getDiscriminant(self):
        return (self.__b ** 2) - (4 * self.__a * self.__c)

    def getRoot1(self):
        discriminant = self.getDiscriminant()
        if discriminant < 0:
            return 0
        return (-self.__b + math.sqrt(discriminant)) / (2 * self.__a)

    def getRoot2(self):
        discriminant = self.getDiscriminant()
        if discriminant < 0:
            return 0
        return (-self.__b - math.sqrt(discriminant)) / (2 * self.__a)


def main():
    print("--- 一元二次方程式求解器 ---")
    try:

        a = float(input("請輸入 a 的值: "))
        b = float(input("請輸入 b 的值: "))
        c = float(input("請輸入 c 的值: "))

        qe = QuadraticEquation(a, b, c)
        disc = qe.getDiscriminant()

        if disc > 0:
            print(f"該方程式有兩個根：{qe.getRoot1():.5f} 與 {qe.getRoot2():.5f}")
        elif disc == 0:
            print(f"該方程式有一個根：{qe.getRoot1():.5f}")
        else:
            print("The equation has no roots")
            
    except ValueError:
        print("輸入錯誤！請輸入數字。")
    except ZeroDivisionError:
        print("錯誤：a 不能為 0，否則這不是一個二次方程式。")

if __name__ == "__main__":
    main()
```

```python
import time

class StopWatch:
    def __init__(self):
        self.__startTime = time.time()
        self.__endTime = self.__startTime

    def get_start_time(self):
        return self.__startTime

    def get_end_time(self):
        return self.__endTime

    def start(self):
        self.__startTime = time.time()

    def stop(self):
        self.__endTime = time.time()

    def getElapsedTime(self):
        return int((self.__endTime - self.__startTime) * 1000)


def main():
    stopwatch = StopWatch()
    
    print("開始計算從 1 加到 1,000,000...")
    
    stopwatch.start()
    
    sum_result = 0
    for i in range(1, 1000001):
        sum_result += i
    
    stopwatch.stop()

    print(f"計算結果: {sum_result}")
    print(f"執行時間: {stopwatch.getElapsedTime()} 毫秒")

if __name__ == "__main__":
    main()
```
